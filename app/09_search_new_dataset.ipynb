### 3.1. NOSIBLE Financial Sentiment

**Kim hazırlamış?**

NOSIBLE Inc ekibi tarafından hazırlanmıştır. Hugging Face veri seti
sayfasında katkı verenler olarak Matthew Dicks, Simon van Dyk ve Stuart Reid
listelenmektedir.

Kaynak: https://huggingface.co/datasets/NOSIBLE/financial-sentiment

**Nasıl oluşturulmuş?**

Veri, NOSIBLE Search Feeds kaynağından örneklenmiştir. Bu kaynak finans
odaklı güvenli sorgulara verilen web arama sonuçlarından oluşur. Tüm veri
public web kaynaklıdır.

Etiketleme şu şekilde yapılmıştır:

- Önce yaklaşık 200 örnek elle etiketlenerek LLM promptları ayarlanmıştır.
- Sonra 100.000 örnek birden fazla LLM annotator ile etiketlenmiştir.
- LLM majority-vote etiketleri embedding tabanlı lineer modellerle kontrol edilmiştir.
- Uyuşmazlık olan örnekler daha büyük bir LLM oracle ile yeniden değerlendirilmiştir.
- Aktif öğrenme tabanlı relabeling döngüsü sonunda final veri seti oluşturulmuştur.

**Uzman anotasyonu var mı?**

Klasik anlamda uzman insan anotasyonlu gold label seti olarak sunulmuyor. Veri
LLM annotatorlar, aktif öğrenme ve LLM oracle relabeling ile oluşturulmuş.
Bu nedenle raw label'lar doğrudan kesin gold label kabul edilmemelidir.

**Bu projede nasıl kullanılabilir?**

- Eğitime doğrudan eklemek yerine yeni dış test havuzu olarak kullanılmalıdır.
- Her sınıftan 100-200 örnek seçilebilir.
- Toplam 300-600 satırlık dengeli bir ek test seti hazırlanabilir.
- Seçilen satırlar insan kontrolünden geçirilerek final label yapılmalıdır.

**Neden iyi aday?**

- Plain sentiment şemasına en kolay uyan aday budur.
- Mevcut Twitter/PhraseBank/S&P 500/Reuters akışından ayrı bir kaynak sunar.
- Hocanın istediği ek veri seti testine en pratik cevap budur.

**Karar**

Birinci tercih olarak önerilir. Ancak insan doğrulaması yapılmadan final
gold label gibi sunulmamalıdır.

### 3.2. SENTiVENT Bench

**Kim hazırlamış?**

SENTiVENT projesi Gilles Jacobs ve ilgili araştırma ekibi tarafından
geliştirilmiştir. Proje ekonomik ve finansal haberlerde event extraction ve
investor sentiment çalışmalarına odaklanır.

Kaynaklar:

- https://jacobsgill.es/sentivent-project
- https://lt3.ugent.be/news/new-sentivent-publications-on-extracting-events-and-sentimen/

**Nasıl oluşturulmuş?**

SENTiVENT, iş/ekonomi haberleri üzerinde olay ve sentiment anotasyonları
oluşturur. Veri setinde ekonomik olay tipleri, olay argümanları ve yatırımcı
sentiment bilgisi birlikte işlenir.

Proje sayfasında event yapısı ve investor sentiment şu şekilde tanımlanır:

- Olaylar belirli bir event typology ile işaretlenir.
- Pozitif, negatif ve nötr investor sentiment ayrı kategori olarak anotasyonlanır.
- Sentiment bazen olayların üzerine bağlanır.
- Örtük sentiment örnekleri özellikle önemlidir.

**Uzman anotasyonu var mı?**

Evet, NOSIBLE'a göre daha güçlü bir manuel/uzman anotasyon karakteri vardır.
SENTiVENT, ekonomik haberlerde event ve sentiment için elle geliştirilmiş anotasyon
şeması ve manuel anotasyon fikriyle sunulur. Bu yönüyle akademik hata analizi
için daha güçlüdür.

**Bu projede nasıl kullanılabilir?**

- Doğrudan ana eğitim verisine eklemek uygun değildir.
- Üç sınıflı plain sentiment şemasına mapping gerekir.
- Daha çok zor/örtük sentiment hata analizi için kullanılabilir.

**Neden iyi aday?**

- Akademik değeri NOSIBLE'dan daha yüksektir.
- Örtük ve olay bazlı sentiment örnekleri tezin tartışma kısmını güçlendirir.

**Risk**

Mevcut proje sade `negative`, `neutral`, `positive` sınıflandırmasına dayanır.
SENTiVENT ise daha karmaşık event-level/aspect-level yapıdadır. Bu nedenle
mapping kuralı iyi yazılmazsa sonuç yorumu tartışmalı olabilir.

**Karar**

İkinci tercih olarak uygundur. Ana ek test yerine gelişmiş hata analizi için
kullanılması daha mantıklıdır.

### 3.3. FNSPID

**Kim hazırlamış?**

FNSPID, Zihan Dong, Xinyu Fan ve Zhiyuan Peng tarafından hazırlanmıştır.
Çalışma `FNSPID: A Comprehensive Financial News Dataset in Time Series`
başlığıyla sunulmuştur ve KDD 2024 Applied Data Science Track ile
ilişkilidir.

Kaynaklar:

- https://github.com/Zdong104/FNSPID_Financial_News_Dataset
- https://huggingface.co/datasets/Zihan1004/FNSPID
- https://arxiv.org/abs/2402.06698

**Nasıl oluşturulmuş?**

FNSPID finans haberleri ile hisse fiyatı zaman serilerini birleştiren büyük
ölçekli bir veri setidir.

Veri seti şunları içerir:

- 15,7 milyon finans haber kaydı.
- 29,7 milyon hisse fiyatı kaydı.
- 4.775 S&P 500 şirketi.
- 1999-2023 dönemi.
- Dört farklı borsa/haber web sitesinden toplanmış haberler.

**Uzman anotasyonu var mı?**

Bu veri seti klasik anlamda uzman insan tarafından etiketlenmiş sentiment gold
seti değildir. Asıl gücü, haberleri tarih, hisse sembolü ve fiyat verisiyle
bağlamasıdır. Bu nedenle sentiment sınıflandırması için doğrudan gold label
olarak değil, piyasa/zaman serisi uzantısı olarak düşünülmelidir.

**Bu projede nasıl kullanılabilir?**

- Tam veri seti kullanılmamalıdır; çok büyüktür.
- Belirli tarih aralığı veya belirli hisseler için küçük örneklem seçilebilir.
- Mevcut fine-tuned modellerle haber başlıklarına sentiment tahmini üretilebilir.
- Daha sonra sentiment ile fiyat yönü arasında basit analiz yapılabilir.

**Neden iyi aday?**

- Tezi piyasa yönü ve zaman serisi tarafına genişletir.
- Haber, hisse sembolü ve fiyat bilgisini aynı yapıda sunar.

**Risk**

- Veri çok büyüktür.
- Doğrudan uzman sentiment etiketi sunmaz.
- Kısa vadeli ek dış test için NOSIBLE kadar pratik değildir.

**Karar**

Üçüncü tercih olarak tutulmalıdır. Ana katkıdan çok ileri çalışma
veya piyasa hareketi analizi için uygundur.
