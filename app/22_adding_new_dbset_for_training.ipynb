```python
!pip install -q datasets

from datasets import load_dataset

dataset = load_dataset("TheFinAI/fiqa-sentiment-classification")

```

```python
import pandas as pd

# Train + test + valid hepsini birleştir
df_fiqa = pd.concat(
    [
        dataset["train"].to_pandas(),
        dataset["test"].to_pandas(),
        dataset["valid"].to_pandas()
    ],
    ignore_index=True
)

# text + label formatına dönüştür
df_fiqa = df_fiqa[["sentence", "score"]].rename(
    columns={
        "sentence": "text",
        "score": "label"
    }
)

# Score -> sentiment label
df_fiqa["label"] = df_fiqa["label"].apply(
    lambda x: "positive" if x > 0
    else "negative" if x < 0
    else "neutral"
)

# Ekrana bas
display(df_fiqa)
```
