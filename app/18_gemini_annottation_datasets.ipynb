```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "app":
    PROJECT_ROOT = PROJECT_ROOT.parent

def read_table(path: Path) -> pd.DataFrame:
    if path.suffix == ".parquet":
        return pd.read_parquet(path)
    if path.suffix == ".csv":
        return pd.read_csv(path)
    if path.suffix in [".xlsx", ".xls"]:
        return pd.read_excel(path)
    raise ValueError(f"Desteklenmeyen dosya türü: {path}")

def inspect_dataset(name: str, path: str, label_col: str, n_rows: int = 3) -> dict:
    file_path = PROJECT_ROOT / path
    df = read_table(file_path)
    
    print("=" * 80)
    print(f"📌 VERİ SETİ: {name}")
    print(f"📁 Dosya Yolu: {path}")
    print(f"📊 Toplam Satır Sayısı (N): {len(df)}")
    print(f"\n🔍 İlk {n_rows} Satır:")
    display(df.head(n_rows))  # Jupyter / Colab kullanmıyorsanız 'display' yerine 'print' kullanabilirsiniz
    print("=" * 80 + "\n")

    # Sınıf sayılarını hesaplama
    counts = df[label_col].astype(str).str.lower().value_counts()

    return {
        "Veri seti": name,
        "N": int(len(df)),
        "Negatif": int(counts.get("negative", 0)),
        "Tarafsız": int(counts.get("neutral", 0)),
        "Pozitif": int(counts.get("positive", 0)),
    }

# Okunacak dosyaların listesi
datasets = [
    ("Eğitim", "db/splits/plain_sentiment_v1/train_df.parquet", "label"),
    ("Doğrulama", "db/splits/plain_sentiment_v1/val_df.parquet", "label"),
    ("Test", "db/splits/plain_sentiment_v1/test_df.parquet", "label"),
    ("S&P 500 dış test", "outputs/finetuned_model_results/finbert_target_finetuned_seed42/external_evaluation/sp500_external/predictions.csv", "gold_label"),
    ("Sentetik test", "outputs/finetuned_model_results/finbert_target_finetuned_seed42/external_evaluation/synthetic_external/predictions.csv", "gold_label"),
    ("Reuters dış test", "outputs/finetuned_model_results/finbert_target_finetuned_seed42/external_evaluation/reuters_external/predictions.csv", "gold_label"),
]

rows = []

# Tüm dosyaları sırayla incele ve ilk 3 satırı bas
for name, path, label_col in datasets:
    row = inspect_dataset(name, path, label_col, n_rows=3)
    rows.append(row)
```
